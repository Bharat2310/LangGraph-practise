```python
from langchain_ollama import ChatOllama 
from langgraph.graph import START, END, StateGraph
from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()
```

```python
model = ChatOllama(model = "gemma3:latest ", temperature=0
)
```

```python
class BlogState(TypedDict):
    title :str
    outline:str
    blog:str
```

```python
def get_outline(state:BlogState)->BlogState:
    title = state["title"]
    prompt = f'give detailed outlined for a blog on topic : {title}'
    result = model.invoke(prompt)
    state["outline"] = result.content
    return state
```

```python
def get_blog(state:BlogState)->BlogState:
    title = state["title"]
    outline = state["outline"]
    prompt = f"write a 1000-1500 words blog on title {title} given outlines : \n{outline}"
    result = model.invoke(prompt)
    state["blog"] = result.content
    return state
    
```

```python

graph = StateGraph(BlogState)
graph.add_node("get_blog", get_blog)

graph.add_node("get_outline", get_outline)


graph.add_edge(START, 'get_outline')
graph.add_edge("get_outline","get_blog" )
graph.add_edge("get_blog", END )




```

```python
workflow = graph.compile()
```

```python
input_state = {"title":"hello word"}
final_state = workflow.invoke(input_state)
```
