```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class PState(TypedDict):
    runs :int
    fours:int
    sixes:int
    sr:float
    balls : int
    b_percent : float
    balls_for_each_boundary : float
    summary : str
```

```python
def Calculate_sr(state: PState)->dict:
    runs = state["runs"]
    balls = state["balls"]
    
    return {"sr": (runs * 100) / balls}
```

```python
def Calculate_b_percent(state: PState)->dict:
    runs = state["runs"]
    fours = state["fours"]
    sixes = state["sixes"]
    
    return {"b_percent": runs / (fours * 4 + sixes * 6)}
```

```python
def calculate_balls_for_each_boundary(state:PState)->dict:
    balls=  state["balls"]
    sixes = state["sixes"]
    fours = state["fours"]
    
    return {"balls_for_each_boundary": balls / (fours + sixes)}
    
```

```python
def summary(state:PState)->dict:
    sr = state["sr"]
    b_percent = state["b_percent"]
    balls_for_each_boundary = state["balls_for_each_boundary"]

    summary_text = f"strike rate : {sr}\n boundary percent:{b_percent}\n balls for each boundary: {balls_for_each_boundary}"
    return {"summary": summary_text}
```

```python
graph = StateGraph(PState)

graph.add_node("Calculate_sr", Calculate_sr)
graph.add_node("Calculate_b_percent", Calculate_b_percent)
graph.add_node("calculate_balls_for_each_boundary", calculate_balls_for_each_boundary)
graph.add_node("summary", summary)

graph.add_edge(START, "Calculate_sr")
graph.add_edge(START, "Calculate_b_percent")
graph.add_edge(START, "calculate_balls_for_each_boundary")
graph.add_edge("Calculate_sr", "summary")
graph.add_edge("Calculate_b_percent", "summary")
graph.add_edge("calculate_balls_for_each_boundary", "summary")
graph.add_edge("summary", END)







```

```python
workflow = graph.compile()
```

```python
initial_state = {"runs":105, "balls":50, "fours":12, "sixes":4}
final_state = workflow.invoke(initial_state)
```
