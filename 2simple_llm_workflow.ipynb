```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama

from dotenv import load_dotenv
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
model = ChatOllama(model = "gemma3:latest ", temperature=1, verbose = True)
```

```python
class LLMState(TypedDict):
    que : str
    ans : str
```

```python
def llm_qa(state:LLMState)->LLMState:
    question = state["que"]
    prompt = f"answer the following question: \n {question}"
    answer = model.invoke(prompt)
    state["ans"] = answer.content
    return state
```

```python
graph = StateGraph(LLMState)
graph.add_node("llm_qa", llm_qa)
graph.add_edge(START, "llm_qa")
graph.add_edge("llm_qa", END)

```

```python
workflow = graph.compile()
```

```python
initial_State = {"que":"what is the distance between sun, moon and earth?"}
final_state = workflow.invoke(initial_State)
```
